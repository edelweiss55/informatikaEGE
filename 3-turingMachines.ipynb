```python
# https://inf-ege.sdamgia.ru/problem?id=83034
def is_prime(n): # функция для проверки числа (простое ли) - ОЧЕНЬ ПОЛЕЗНО!!!!
    if n < 2:
        return False
    for i in range(2, int(n**0.5)+1): #for i in range(n+1)
        if n % i == 0:
            return False
    return True


    
total = 1200
for zeros in range(total, 1, -2):# Перебираем чётные нули от максимума вниз
    sum_ones = total - zeros# исходные единицы
    if is_prime(sum_ones):# проверяем условие простоты
        print(zeros)
        break
```
