# Библиотека itertools и комбинаторные задачи

## Что такое itertools?

`itertools` - стандартная библиотека Python, облегающая работу с коллекциями, например: инициализацию с заполнением нужными данными, обход, различные преобразования (перестановки, замены, разворот в обратном порядке и т.д.). За счет грамотного и умелого использования этой библиотеки можно многократно ускорить решение различных задач и уменьшить объем написанного кода, так как громоздкие циклы и самописаные функции по обходу и преобразованиям заменяются стандартными функциями этой библиотеки. 

## Что такое комбинаторика?

Комбинаторика - раздел математики, посвященный способам выбора, перестановки и сочетаний элементов в соответствии с заданными условиями. В рамках задач ЕГЭ (по информатике, да и в принципе, и в ЕГЭ по профильной математике может выпасть какое-то задание с формулами по комбинаторике - те же задачи по теорверу или последнее задание второй части по теории чисел) мы будем часто сталкиваться с потребностью выявить, сколько существует возможных комбинаций тех или иных объектов: сколько номеров телефона/IP-адресов/слов из N букв/программ для машины Тьюринга с такими-то условиями можно составить, сколько номеров телефона/IP-адресов/слов из N букв/программ для машины Тьюринга находится между таким-то и вот таким-то вариантом, какой по счету среди номеров телефона/IP-адресов/слов из N букв/программ для машины Тьюринга такой-то вариант и т.д. При хорошей математической базе и всего паре-тройке комбинаторных формул, которые отложились в долгосрочной памяти, можно решать подобные типы задач вручную (без разницы: в уме или на черновике), без применения Python, как вспомогательного инструмента, но лучше уметь и так, и эдак) 

## Основные комбинаторные понятия

### Перестановки 

Сразу запомните его английский перевод: **permutations** - так будет проще вспоминать о них в рамках программирования. Перестановкии - это упорядочивание *n* элементов какого-либо множества, порядок важен, задействованы все *n* элементов. Грубо говоря - это тасовка, распределение каждого элемента по местам. Представьте, что вы - администратор отеля, в котором есть ровно 50 пустующих номеров, к вам заезжают одномоментно 50 клиентов, и ваша задача - расселить каждого по этим 50 номерам (по одному человеку в номер!). Количество всевозможных вариантов расселения - и есть количество всевозможных перестановок. Та же задача нависает над нами, когда нам надо разложить все игральные карты из колоды в ряд, отсортировать учеников класса в шеренгу на уроке физкультуры, определить порядок выполнения дел в своем todo-листе (если вы практикуете тайм-менеджмент) и так далее. Главное, что нужно учитывать, когда мы говорим о перестановках: 

1) Всегда берем все *n* элементов - то есть, мы не можем какие-то элементы отбросить, предложив не заселять номера отеля 38 и 40, выгнав на улицу двух потенциальных постояльцев, или же оставив в карточной колоде только карты одной колоды, потому что другие нам не нравятся, в любой комбинации перестановок задействованы **абсолютно** все элементы выборки!
2) Все рассматриваемые элементы уникальны и отличимы друг от друга - среди клиентов нашего отеля не будет клонов, а в нашей карточной колоде нет дубликатов (иначе это уже шулерство), даже если объекты кажутся одинаковыми, они должны быть пронумерованы, чтобы мы отличали их друг от друга. Если же от нас требуется найти количество перестановок для *n* белых и *m* черных шаров/шашек/голубей, то просто найти количество перестановок нельзя, ведь тут нам уже все равно, в каком порядке будут расположены три абсолютно неотличимых объекта, если они оказались рядом друг с другом в какой-либо комбинации - эти комбинации для нас тоже будут неотличимыми.

Чисто всех перестановок для *n* элементов запоминается по такой вот формуле:

$P_n = n!$

Где *n!* - факториал от натурального числа *n*, то есть - произведение всех чисел последовательно от 1 до n включительно. Пример:
$10!=1\cdot2\cdot3\cdot4\cdot5\cdot6\cdot7\cdot8\cdot9\cdot10$

Как это понять? Представьте себе тот же отель, но сократим количество пустующих комнат и бездомных клиентов до 5. Какими способами можно их сопоставить? Представьте эти пять номеров последовательными дверьми слева направо и пройдитесь слева направо, начиная с первого номера, постояльцы нервно стоят прямо перед глазами, пока что все пятеро. Сколько вариантов постояльцев можно заселить в первый номер? Пять. Одного постояльца уже определили, переходим правее, ко второму номеру, ещё четыре человека топчутся в коридоре. А сюда теперь сколько останется вариантов для заселения? Четыре. Вот уже 20 вариантов для заселения первых двух номеров (рассмотрим каждый из пяти вариантов для первого номера, на второй во всех пяти случаях останется четыре возможных жильца - итого вариантов двадцать). Для третьего - три варианта, для четвёртого - уже два, ну и пятый номер заселяется автоматически последним оставшимся клиентом. Перемножаем пять на четыре на три на два на один - вот и получается факториал от пяти (посчитайте, сколько в итоге вариантов заселения будет для всех пятерых, если принцип стал понятен - замахнитесь на изначальный вариант из пятидесяти номеров и жильцов).

### Размещения

Тут же запомните английский перевод **arrangments**. Размещения - комбинации, связанные с выбором *k* элементов из *n* вариантов. В перестановках, как вы помните, из *n* вариантов мы всегда берём все *n* - тут без вариантов)
Представим, что в нашем отеле теперь 50 потенциальных клиентов, но всего 10 свободных номеров - теперь мы можем устроить кастинг, отобрав 10 самых пристойных кандидатов на заселение. Порядок размещения по-прежнему важен - номера в отеле же пронумерованы (логично). 
Вариантов для отбора у нас:
$A^k_n=\frac{n!}{(n-k)!}$
Давайте подумаем, как это объяснить и вывести путем рассуждения, если вдруг забудете формулу. 
Числитель, как можно заметить, представляет собой число перестановок всех *n* элементов - мы уже знакомы с этим концептом. В знаменателе, с некоторой точки зрения, тоже скрылось число перестановок *n-k* элементов - то есть того, что останется после нашего отбора (всех бездомных, кому номер в отеле не достанется). Попытаемся по этой формуле теперь восстановить мысленно процесс отбора наших размещений: сначала мы получаем число перестановок, тасуем, всевозможные варианты (претендентов для отбора), отсюда и *n!*; затем первые *k* позиций отбираем; останется "за бортом" *n-k* бедолаг - их порядок или способ перестановки нам уже неважен, потому *(n-k)!* способов их тасовки мы откидываем за ненадобностью - в нашем же случае делим на них число перестановки всех элементов.

#### Размещения с повторениями

Размещения с повторениями отличаются от классических тем, что элементы можно отбирать несколько раз повторно, хоть *k* раз подряд выбрать один и тот же элемент среди *n*. Потому формула в разы проще:
$\overline{A^k_n}=n^k$
То есть: на каждом шаге отбора очередного кандидата у нас всегда *n* доступных вариантов (претендентов). Представьте, что теперь нашим заселенным постояльцам отеля нужно раздарить *k* купонов на бесплатное посещение отельного бара. Можем по честному раздать их *k* самым культурным гражданам, можем сделать рандомизированный розыгрыш, а можем скинуть все *k* купонов своему другу, который специально ради них заселился в наш отель на неделю. Но во всяком случае у нас $n^k$ вариантов, как распорядиться купонами. 

### Сочетания

Теперь представим, что нам нужно посчитать количество размещений, но теперь объекты неразличимы, либо же нам просто уже неважен порядок или нумерация отобранных объектов - в этом случае мы столкнемся с *сочетаниями*. Ну и английский вариант - **combinations**. Представьте, что теперь из этих *n* постояльцев отеля надо выселить *k* - нам совершенно неважно, в каком порядке их выгонять, в какую шеренгу они выстроятся перед выходом, как будут стоять относительно друг друга и т.д. Сколько способов освободить номера у нас в распоряжении?
$C^k_n=\frac{n!}{(n-k)!k!}$


От размещений эта формула отличается только дополнительным делением на *k!* - число перестановок *k* изгнанных из нашего отеля, то есть, как раз-таки мы убираем из расчета всевозможные способы тасовки этих граждан. 

## Основы itertools

itertools оперирует такими объектами, как **итераторы**, которые позволяют перебирать последовательности элементов, выдавая их по одному и запоминая текущую позицию.

itertools предоставляет три основных типа итераторов:

1) Бесконечные итераторы
2) Конечные итераторы
3) Комбинаторные итераторы

```python
import itertools
```

### Бесконечные итераторы

Предназначены для бесконечного перебора, например, последовательного счета или цикличного прохода по списку. Остановить можно только вручную, контролируя счетчики, проверяя какое-либо условие и т.д. Либо же можно оставить бесконечный проход итератора, если, например, нужно показывать какую-то анимацию или другой бесконечный процесс. 

Бесконечных итераторов всего три: `count()`, `cycle()`, `repeat()`, они в равной мере могут пригодиться при решении некоторых ЕГЭшных задач.

#### `count()`

`count(start=0, step=1)` представляет собой бесконечный аналог `range`, позволяющий вести бесконечный последовательный счёт, начиная со `start` и с шагом `step`. В ЕГЭ-задачах такой итератор может пригодиться там, где заранее неизвестна верхняя граница перебора, либо мы хотим быстро обойти какой-то диапазон с постоянным шагом.


```python
# посчитаем от 0 до 10
for i in itertools.count(start=0, step=1):
    if i > 10:
        break
    print(i)
```

```python
# формируем список из равноудаленных друг от друга значений
levels = list(itertools.islice(itertools.count(0, 10), 5))
print(levels)
```

```python
# теперь представим, что levels - уровни громкости, которые мы можем выбирать (например, для телевизора)
# сопоставим значения уровней с их словесным описанием
level_names = ["mute", "тихо-тихо", "тихо", "нормально", "громко", "вообще громко кошмар"]

level_map = dict(zip(itertools.count(0, 10), level_names))

print(level_map)
```

#### `cycle()`
`cycle(iterable)` принимает в качестве аргумента какой-то итерируемый объект - список или строку, чтобы ее можно было бесконечно обходить "по кругу" - отсюда и название *cycle*. 

```python
# равномерно распределяем задачи по кругу 

from itertools import cycle

tasks = ['Уборка', 'Ознакомление с GTA VI', 'Готовка', 'Кушать чипсы', 'Стирка', 'Прогулка', 'Готовиться к ЕГЭ']
people = ['Анна', 'Борис']

schedule = list(zip(tasks, cycle(people)))
for task, person in schedule:
    print(f"{task} -> {person}")
```

#### `repeat()`

`repeat(elem, n)` повторяет `elem` `n` раз (или бесконечно, если второй аргумент не передать). 

```python
# быстрое заполнение значениями по умолчанию
users = ["ivan", "anna", "oleg"]
user_statuses = dict(zip(users, itertools.repeat("готовится к ЕГЭ")))
user_statuses["ivan"] = "сдал ЕГЭ"
print(user_statuses)
```

```python
# быстрое возведение набора чисел в куб
cubes = list(map(pow, [2, 3, 4], itertools.repeat(3)))
print(cubes)
```

## Конечные итераторы

Предназначены для разового прохода по спискам - то есть, это не какая-то зацикленная история, которая может повториться по кругу, как в случае бесконечных итераторов. С ними нам не надо вручную проверять условия и контролировать завершение обхода. 

### Обработка и фильтрация данных

#### `accumulate()`

Вовзращает накопленные суммы, либо результат другой бинарной (от двух аргументов) функции.

```python
# из последовательности начислений сделаем список обновлений баланса
from itertools import accumulate
import operator

payrolls = [100, 500, -300, 1000, 500]
print(list(accumulate(payrolls)))

# а тут попробуем то же, но с умножением - подсчитаем общий процент начислений (что-то похожее мы иногда считаем в экономической задаче ЕГЭ по математике)
accruals = [1.25, 1.05, 1.02, 1.05, 1.15]
print(list(accumulate(accruals, operator.mul)))
```

#### `filterfalse()`

`filterfalse(predicate, iterable)` - возвращает элементы, для которых `predicate` ложен.

```python
# выводим только нечетные
from itertools import filterfalse 

li = [2, 4, 5, 7, 8, 10, 20]  
print (list(itertools.filterfalse(lambda x : x % 2 == 0, li))) 
```

#### `compress()`
`compress(data, selectors)` - оставляет только те элементы data, которым в selectors соответствуют `1` или `True`.

```python
# отберем только тех, кто набрал высокие баллы по ЕГЭ
from itertools import compress

users = ['Анна', 'Борис', 'Олег', 'Мария']
scores = [85, 42, 91, 60]

# Создаем маску: оставляем тех, у кого балл выше 70
passed_mask = [score > 70 for score in scores]

top_users = list(compress(users, passed_mask))
print(top_users)
```

#### `takewhile()`
`takewhile(predicate, iterable)` возвращает элементы до тех пор, пока соблюдается условие из `predicate`.

```python
from itertools import takewhile

numbers = [2, 4, 6, 7, 8, 10]

# Берем числа, пока они четные
result = list(takewhile(lambda x: x % 2 == 0, numbers))
print(result)
```

#### `dropwhile()`
`dropwhile(predicate, iterable)` - пропускает элементы, пока условие истинно, при первом же `False` выдаст все оставшиеся элементы (независимо от истинности предиката для них).

```python
from itertools import dropwhile

numbers = [2, 3, 4, 5, 6, 7, 8]

# Пропускаем числа, пока они меньше 5
result = list(dropwhile(lambda x: x < 5, numbers))
print(result)
```

### Объединение и группировка

#### `chain()`
`chain(*iterables)` - "склеивает" несколько последовательностей в одну, но в отличие от сложения списков через `+, копирования не происходит, итератор перебирает последовательности без выделения доп.памяти.

```python
from itertools import chain

list_a = [1, 2, 3]
tuple_b = ('a', 'b')
str_c = "XY"

combined = list(chain(list_a, tuple_b, str_c))
print(combined)
```

#### `zip_longest()`

`zip_longest(*iterables, fillvalue)` - в отличие от обычного `zip` объединение происходит по большей последовательности, все недостающие элементы заполняются значением по умолчанию `fillvalue`.

```python
from itertools import zip_longest

names = ['Анна', 'Борис', 'Олег', 'Мария', 'Иксат']
scores = [98, 40, 60, 89]

# Недостающие баллы заполняем нулями
paired = dict(zip_longest(names, scores, fillvalue=0))
print(paired)
```

### Срезы, разбиения

#### `pairwise()`

`pairwise(iterable)` - дробит последовательность по перекрывающимся парам.

```python
from itertools import pairwise

letters = ['A', 'B', 'C', 'D', 'E']

pairs = list(pairwise(letters))
print(pairs)
```

#### `batched()`
`batched(iterable, n)` - по сути, тоже дробит на пары, но теперь можно настроить длину каждого такого дробления (делать уже не по два элемента, а, например, по три).

```python
from itertools import batched

items = ['A', 'B', 'C', 'D', 'E', 'F', 'G']

# Разбиваем на группы по 3 элемента
chunks = list(batched(items, 3))
print(chunks) # G остался без продолжения - последний кортеж по длине может оказаться короче, чем n
```

### Комбинаторные итераторы

Как раз призваны для перебора элементов в классических комбинаторных задачах, рассмотренных выше, когда мы обсуждали математический аспект темы. 

#### `product()`
`product(iterables, repeat=1)` - декартово произведение коллекций (всевозможное разбиение по парам), зачастую помогает заменить вложенные циклы.

```python
# перечислим всевозможные конфигурации футболок
from itertools import product

colors = ['Красный', 'Синий', 'Желтый', 'Черный', 'Белый']
sizes = ['S', 'M', 'L', 'XXXL']

# Все возможные варианты товаров (аналог вложенного цикла for)
variants = list(product(colors, sizes))
print(variants)
```

```python
# сколько пин-кодов можно составить из цифр '1', '2', '3', '4'?
pin_codes = list(product([1,2,3,4], repeat=4))
print(len(pin_codes)) # попробуйте без len посмотреть все эти комбинации
```

#### `permutations()`

`permutations(iterable, r=None)` - перестановки из объектов `iterable` длиной `r` (по дефолту это самая классическая перестановка из всех элементов). 

```python
from itertools import permutations

items = ['A', 'B', 'C']

# Все возможные способы расставить 3 элемента (3! = 6 вариантов)
all_perms = list(permutations(items))
print(all_perms)
```

```python
# Размещения

from itertools import permutations

items = ['A', 'B', 'C']

# Все возможные способы расставить 3 элемента (3! = 6 вариантов)
all_perms = list(permutations(items, r=2))
print(all_perms)
```

#### `combinations()`

`combinations(iterable, r)` - сочетания из `iterable` по `r` элементов. 

```python
from itertools import combinations

players = ['Анна', 'Иван', 'Олег', 'Мария']

# Все возможные уникальные пары игроков для турнира (без дублей вида Иван-Анна)
pairs = list(combinations(players, 2))
print(pairs)
```

#### `combinations_with_replacement()`

```python
# Пример условия: Сколько различных слов можно составить перестановкой букв слова КАПКАН так, 
# чтобы две одинаковые буквы не стояли рядом?

from itertools import permutations

valid_words = set()  # Множество автоматически уберет дубликаты

for p in permutations('КАПКАН'):
    word = ''.join(p)
    # Проверяем, что одинаковые буквы не стоят рядом
    if 'КК' not in word and 'АА' not in word:
        valid_words.add(word)

print(len(valid_words))
```

```python
# Построение таблицы истинности с помощью itertools
from itertools import product

print("x y z w")
# Генерируем 16 наборов из нулей и единиц для 4 переменных
for x, y, z, w in product([0, 1], repeat=4):
    # Подставляем выражение из условия задачи (пример: (x <= y) and (y != z) and not w)
    f = (x <= y) and (y != z) and not w
    if f:  # Или if not f:, если в таблице даны нули
        print(x, y, z, w)
```
