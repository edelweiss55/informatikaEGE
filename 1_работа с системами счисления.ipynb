# Работа с системами счисления

## Перевод из десятичной системы счисления в другие

Для перевода десятичного числа в одну из общепринятых систем счисления (двоичную, восьмеричную и шестнадцатиричную) используются стандартные функции, приведенные ниже. Все они на вход получают **число**, а на выход отдают **строку**.

#### *Перевод в двоичную систему счисления*

Перевод числа в бинарную систему счисления из десятичной осуществляется функцией `bin(num)`

#### *Перевод в восьмеричную систему счисления*

Перевод числа в восьмеричную систему счисления из десятичной осуществляется функцией `oct(num)`

#### *Перевод в шестнадцатиричную систему счисления*

Перевод числа в шестнадцатиричную систему счисления из десятичной осуществляется функцией `hex(num)`

```python
x = 267
print('x в десятичной: ', x)
xbin = bin(x)
print('x в двоичной: ', xbin)
xoct = oct(x)
print('x в восьмеричной: ', xoct)
xhex = hex(x)
print('x в шестнадцатиричной: ', xhex)
```

### Как убрать префикс в начале?

При выводе числа в другой системе счисления первые два символа - префикс, обозначающий систему счисления. `0b` - двоичная; `0o` - восьмеричная; `0x` - шестнадцатиричная. Чтобы просто получить само представление числа без этого префикса, необходимо его обрезать:

`xbinWithoutPrefix=xbin[2:]`

`[2:]` - получение среза (slice), такая запись обозначает: обрезать первые два элемента массива и оставить все остальное (срез от третьего элемента включительно и до конца)

```python
xbinWithoutPrefix=xbin[2:]
print('x в двоичной без префикса: ', xbinWithoutPrefix)
```

### Работа со слайсами

- `[x:]` - обрезать первые x элементов, остальное оставить
- `[:x]` - оставить первые x элементов, остальное обрезать
- `[-x:]` - оставить только последние x элементов, остальное обрезать
- `[:-x]` - обрезать последние x элементов, остальное оставить
- `[a:b]` - срез от a (включительно) до b (не включительно)

***Внимание! Массивы начинают считаться с нулевого индекса!***

```python
a='123456789'
print(a[2:])
print(a[:2])
print(a[-2:])
print(a[:-2])
print(a[2:5])
```

## Перевод из других систем счисления в десятичную

Для перевода числа обратно в десятичную систему из любой другой используется функция `int(x, pos)`, где `x` - строка с записью числа в исходной системе счисления, а `pos` - основание этой системы (2 для двоичной, 8 для восьмеричной, 3 для троичной и т.д.). Результатом работы функции будет **число**.

```python
print(int(xhex, 16))
```

```python
# попробуем с произвольной строкой:

print(int('bebebeb', 8)
```

```python
# попробуем и так:

print(int('12345', 3))
```

```python
# и вот так:

print(int('FFA', 16))

# и вот так:

print(int('0xFFA', 16))
```

## Перевод из десятичной системы в любую нестандартную

Стандартных функций, к сожалению нет, поэтому нужно вручную получать остатки от деления числа на основание новой системы счисления
с последующим делением числа, чтобы убирать уже посчитанные разряды, проще всего разобрать перевод в троичную систему:

```python
def f(n):
    s='' # результат, куда будем вписывать разряды
    # делим, пока число не закончится, грубо говоря:
    while n > 0: 
        s = str(n%3)+s # делим на основание системы (3), прибавляем новый разряд к результату (почему не s+str(n%3)?)
        n //= 3 # делим нацело на 3, чтобы убрать младший, уже записанный разряд (почему нельзя делить обычным образом?)
    return s

for i in range(1, 20):
    print(f(i))
```

Алгоритм получает на вход натуральное число N и строит по нему новое число R следующим образом.
- 1.  Строится троичная запись числа N.

- 3.  Результат переводится в десятичную систему счисления.
- 4.  Результатом работы алгоритма становится модуль разности исходного числа N и числа, полученного на предыдущем шаге.
 
Пример. Дано число N  =  35. Алгоритм работает следующим образом.
- 1.  Строим троичную запись числа N: 3510  =  10223.
- 2.  Заменяем цифры и удаляем ведущие нули: 1022 → 1200.
- 3.  Переводим в десятичную систему: 12003  =  4510.
- 4.  Вычисляем модуль разности: |35 − 45|  =  10.
  
Результат работы алгоритма R  =  10.
При каком наименьшем N в результате работы алгоритма получится R  =  1 864 246?

```python
for n in range(1, 10000000):
    ntr = f(n) # троичная запись числа
    ntrreplace = ntr.replace('0', '_').replace('2', '0').replace('_', '2') # почему нужна такая замена ('0', '_')?
    resDes = int(ntrreplace, 3)
    res = abs(resDes - n)
    if res == 1864246:
        print(n)
        break
```
